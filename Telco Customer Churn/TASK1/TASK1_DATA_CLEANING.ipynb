```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
print(os.getcwd())
```

```python
df = pd.read_csv('Telco_Customer_Churn_Dataset.csv')
df.head()
```

```python
print("Rows and Columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
print("Unique error values in TotalChargers Column:")
print(df[df['TotalCharges'] == ' ']['TotalCharges'].count())
```

```python
print(df[df['TotalCharges'] == ' '][['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', '0')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
print("TotalCharges error has been fixed! Here is what's been fixed:")
print(df[df['tenure'] == 0][['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']])
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
print("Churn column converted:")
print(df['Churn'].value_counts())
print("\nSample:")
print(df['Churn'].head(10))
```

```python
df_encoded = pd.get_dummies(df, columns=['Contract', 'PaymentMethod', 
                                          'InternetService', 'gender',
                                          'Partner', 'Dependents',
                                          'PhoneService', 'PaperlessBilling'],
                             drop_first=True)
print("Original columns: 21")
print("New shape after encoding:", df_encoded.shape)
print("\nNew column names:")
print(df_encoded.columns.tolist())
```

```python
df_encoded.to_csv('Cleaned_Churn_Data.csv', index=False)

print("Cleaned data saved as: 'Cleaned_Churn_Data.csv'")
print("Task 1 Complete:")
print(f"\nFinal dataset: {df_encoded.shape[0]} customers, {df_encoded.shape[1]} columns")
```
